```python
import os
import numpy as np
import pandas as pd
import yfinance as yf


import matplotlib.dates as mdates
import matplotlib.pyplot as plt

import plotly.express as px

```

```python
#Descargar datos mensuales directamente desde Yahoo Finance

start_date = "2020-01-01"
end_date = "2021-12-31"

# Descargar precios mensuales ajustados automáticamente
merval = yf.download('^MERV', start=start_date, end=end_date, interval='1d', auto_adjust=True)

    
# Resetear índice y quedarnos con columnas necesarias
merval = merval[['Close']].reset_index()


# Renombrar columnas
merval.columns = ['Date', 'Merval']

```

```python
df_merval = pd.DataFrame(merval)
df_merval.info()
```

```python
# Agrupar por fecha y contar ids

fig = px.line(
    df_merval,
    x="Date",
    y="Merval",
    markers=True,
    title="Cantidad de IDs por fecha"
)

fig.update_layout(
    xaxis_title="Fecha",
    yaxis_title="Cantidad de IDs",
    xaxis=dict(tickangle=45)
)

fig.show()
```
