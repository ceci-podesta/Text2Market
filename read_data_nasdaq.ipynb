# Armado del subdataset para el siguiente step del proyecto

Por el momento, nos vamos a quedar únicamente con los registros que tengan datos no nulos en las columnas 'Article_title', 'Stock_symbol', 'Article' y 'Url' ya que, a priori, son las que vamos a utilizar para el análisis de sentimiento y la predicción de precios.

Adicionalmente, vamos a eliminar las columnas con los summary (si bien esto podría ser útil, tomamos la decisión de trabajar en el proyecto únicamente con los datos de texto originales procesados por nosotras, emulando lo que a futuro deberíamos realizar si quisieramos construir un pipeline que replique lo que estamos haciendo en el proyecto con noticias obtenidas vía scrapping).

Por último, vamos a eliminar la columna 'Publisher' por la elevada cantidad de valores nulos y la columna Unnamed: 0 que es un índice innecesario.

```python
df_proyecto = df[df[['Article_title', 'Stock_symbol', 'Article', 'Url']].notna().all(axis=1)].copy()
```

```python
df_proyecto.drop(columns=['Unnamed: 0', 'Lsa_summary', 'Luhn_summary',
       'Textrank_summary', 'Lexrank_summary','Publisher'], inplace=True)
```

```python
df_proyecto.to_csv('nasdaq_subdataset.csv', index=False)
```
