```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

# Read s300news_summary

```python
folder = 'data\hs300news_summary'
dfs = []

for file in os.listdir(folder):
    if file.endswith('.jsonl'):
        path = os.path.join(folder, file)
        df = pd.read_json(path, lines=True)
        dfs.append(df)

# Combinar todo en un solo DataFrame
df_news = pd.concat(dfs, ignore_index=True)
```

```python
print(df_news.shape)
print(df_news.info())
df_news.head()
```

```python
df_news.title.value_counts().to_csv('outputs/news_title_counts.csv', header=['count'])
```

```python
# df_sp500: tu sub-DataFrame filtrado y con Date tipo datetime
df_sp500['Date'] = pd.to_datetime(df_sp500['Date'], errors='coerce')

# Filtrar años 2000-2026
df_sp500 = df_sp500[(df_sp500['Date'].dt.year >= 2000) & (df_sp500['Date'].dt.year <= 2026)]

# Contar artículos por día
articulos_por_dia = df_sp500.groupby('Date').size()

# Crear índice de fechas completo
fecha_inicio = df_sp500['Date'].min()
fecha_fin = df_sp500['Date'].max()
fechas_completas = pd.date_range(start=fecha_inicio, end=fecha_fin, freq='D')

# Reindexar para que los días sin artículos aparezcan con 0
articulos_por_dia = articulos_por_dia.reindex(fechas_completas, fill_value=0)

# Pasar a DataFrame
df_dias = articulos_por_dia.reset_index()
df_dias.columns = ['Date', 'count']

# Crear columna Año-Mes como string
df_dias['YearMonth'] = df_dias['Date'].dt.strftime('%Y-%m')

# Contar días sin artículos por mes, incluyendo meses sin huecos
dias_sin_articulos = df_dias.groupby('YearMonth').apply(lambda x: (x['count'] == 0).sum())

# Mostrar resultados
print(dias_sin_articulos)
```

```python
dias_sin_articulos.to_csv('outputs/sp500_dias_sin_articulos.csv', header=['Dias_sin_articulos'])
```

```python
# df_clean: tu sub-DataFrame filtrado con las 3 columnas no nulas
df_sp500['Date'] = pd.to_datetime(df_sp500['Date'], errors='coerce')
df_sp500 = df_sp500[(df_sp500['Date'].dt.year >= 2015) & (df_sp500['Date'].dt.year <= 2026)]

# Contar artículos por día
articulos_por_dia = df_sp500.groupby('Date').size()

# Crear DataFrame con año
df_dias = articulos_por_dia.reset_index()
df_dias.columns = ['Date', 'count']
df_dias['Year'] = df_dias['Date'].dt.year

# Configuración de subplots: 13 filas x 1 columnas
fig, axes = plt.subplots(nrows=13, ncols=1, figsize=(20, 80), sharex=False, sharey=False)
axes = axes.flatten()

years = range(2015, 2026)
for i, year in enumerate(years):
    ax = axes[i]
    data_year = df_dias[df_dias['Year'] == year]['count']
    
    # Histograma
    ax.hist(data_year, bins=30, color='skyblue', edgecolor='black')
    
    # Contar total de artículos en ese año
    total_articulos = data_year.sum()
    
    # Título con año + total de artículos
    ax.set_title(f'Año {year} (Artículos: {total_articulos})')
    ax.set_xlabel('Artículos por día')
    ax.set_ylabel('Días')
    
# Ocultar subplots sobrantes
for j in range(len(years), len(axes)):
    axes[j].axis('off')

plt.tight_layout()
plt.show()
```
